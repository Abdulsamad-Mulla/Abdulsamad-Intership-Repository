```python
# 1) Total Transaction
total_transaction = len(df)
print("Total Transaction : ", total_transaction)
```

```python
# 2) Total Fraud Transaction
fraud_transactions = df["fraud"].sum()
print("Fraud Transaction :", fraud_transactions)
```

```python
# 3) Fraud Percentage
fraud_transaction = (fraud_transactions / total_transaction) * 100
print("Fraud Percentage :", fraud_transaction)
```

```python
# VISUALIZATION
```

```python
# 1. Fraud vs Genuien (Count Plot)

fraud_counts = df["fraud"].value_counts().sort_index()

plt.figure(figsize=(6,5))
plt.bar(["Genuien", "Fraud"], fraud_counts)

plt.title("Fraud Vs Genuien Transaction")
plt.xlabel("Transaction Type")
plt.ylabel("Count")

plt.show()
```

```python
# 2. TRANSACTION DISTRIBUTION

plt.figure(figsize=(8,5))

plt.hist(df["ratio_to_median_purchase_price"],
         bins=30,
         edgecolor="black")

plt.title("Transaction Distribution")
plt.xlabel("Ratio to Median Purchase Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# 3. BOX PLOT
plt.figure(figsize=(8,5))

plt.boxplot(df["ratio_to_median_purchase_price"],
vert=False)

plt.title("Box plot of transaction ratio")
plt.xlabel("Ratio to Median Purchase Price")

plt.show()
```

```python
# 4. Distribution Plot
plt.figure(figsize=(8,5))

plt.hist(df["ratio_to_median_purchase_price"],
bins=50,
density=True,
edgecolor="Black")

plt.title("Distribution of Transaction Ration")
plt.xlabel("Ration to Median Purchase Price")
plt.ylabel("Density")

plt.show()
```

```python
# 5. Fraud by Online Order
online_fraud = pd.crosstab(df["online_order"], df["fraud"])

online_fraud.plot(kind="bar", figsize=(7,5))

plt.title("Fraud by Online Order")
plt.xlabel("Online Order")
plt.ylabel("Number of Transaction")
plt.xticks([0,1], ["Offline", "Online"], rotation=0)
plt.legend(["Genuine", "Fraud"])

plt.show()
```

```python
# 6. FRAUD BY USED CHIP

chip_fraud = pd.crosstab(df["used_chip"], df["fraud"])
chip_fraud.plot(kind="bar" , figsize=(7,5))

plt.title("Fraud By Used Chip")
plt.xlabel("Used Chip")
plt.ylabel("Number Of Transaction")
plt.xticks([0,1] , ["No", "Yes"], rotation=0)
plt.legend(["Genuien", "Fraud"])

plt.show()
```

```python
# 8. Fraud By Used Pin

pin_fraud = pd.crosstab(df["used_pin_number"], df["fraud"])
pin_fraud.plot(kind="bar" , figsize=(7,5))

plt.title("Fraud By Used pin")
plt.xlabel("Used pin")
plt.ylabel("Number Of Transaction")
plt.xticks([0,1] , ["No", "Yes"], rotation=0)
plt.legend(["Genuien", "Fraud"])

plt.show()
```

```python
# 9. Fraud by Repeat Retailer

repeat_fraud = pd.crosstab(df["repeat_retailer"], df["fraud"])
repeat_fraud.plot(kind="bar" , figsize=(7,5))

plt.title("Fraud By Repeat Retailer")
plt.xlabel("Repeat Retailer")
plt.ylabel("Number Of Transaction")
plt.xticks([0,1] , ["No", "Yes"], rotation=0)
plt.legend(["Genuien", "Fraud"])

plt.show()
```

```python
# 10. Scatter Plot

plt.figure(figsize=(8,6))

plt.scatter(df["distance_from_home"],
            df["ratio_to_median_purchase_price"],
            alpha=0.3)

plt.title("Distance from Home vs Purchase Ratio")
plt.xlabel("Distance from Home")
plt.ylabel("Ratio to Median Purchase Price")

plt.show()
```

```python
# 11. Correlation Heatmap 

corr = df.corr(numeric_only=True)

plt.figure(figsize=(8,6))

plt.imshow(corr, cmap="coolwarm", interpolation="nearest")

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```
