# Day 3 — Exploratory Data Analysis (EDA)

Analyzing the job dataset: counts, common roles, locations, description length, missing values, duplicates, and 5 visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/jobs_skills_Extraction_10000.csv')
df['description_length'] = df['job_description'].str.len()
print('Number of jobs:', len(df))
print('Unique job titles:', df['job_title'].nunique())
```

## Most common job roles

```python
top_titles = df['job_title'].value_counts().head(10)
top_titles
```

## Jobs by location

```python
top_locations = df['location'].value_counts().head(10)
top_locations
```

## Description length stats

```python
df['description_length'].describe()
```

## Missing values & duplicates

```python
print(df.isnull().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## Top companies

```python
top_companies = df['company'].value_counts().head(10)
top_companies
```

## Visualization 1: Top job titles

```python
plt.figure(figsize=(8,5))
top_titles.plot(kind='barh', color='#4C72B0')
plt.title('Top 10 Job Titles')
plt.xlabel('Count')
gca=plt.gca()
gca.invert_yaxis()
plt.tight_layout()
plt.savefig('viz_top_titles.png', dpi=120)
plt.show()
```

## Visualization 2: Jobs by location

```python
plt.figure(figsize=(8,5))
top_locations.plot(kind='barh', color='#55A868')
plt.title('Top 10 Locations')
plt.xlabel('Count')
gca=plt.gca()
gca.invert_yaxis()
plt.tight_layout()
plt.savefig('viz_locations.png', dpi=120)
plt.show()
```

## Visualization 3: Description length distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['description_length'], bins=30, color='#C44E52')
plt.title('Description Length Distribution')
plt.xlabel('Characters')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('viz_desc_length.png', dpi=120)
plt.show()
```

## Visualization 4: Missing values

```python
miss = df.isnull().sum()
miss = miss[miss > 0]
plt.figure(figsize=(6,4))
miss.plot(kind='bar', color='#8172B2')
plt.title('Missing Values by Column')
plt.ylabel('Missing count')
plt.tight_layout()
plt.savefig('viz_missing.png', dpi=120)
plt.show()
```

## Visualization 5: Top companies

```python
plt.figure(figsize=(8,5))
top_companies.plot(kind='barh', color='#CCB974')
plt.title('Top 10 Companies')
plt.xlabel('Count')
gca=plt.gca()
gca.invert_yaxis()
plt.tight_layout()
plt.savefig('viz_companies.png', dpi=120)
plt.show()
```

## Summary
- 10,000 job postings across 20 unique job titles
- Reasonably even split across job types and locations (synthetic/generated dataset)
- Description lengths are consistent (low variance), suggesting template-generated postings
- Missing values are limited to `company`, `education`, `salary` — none in `job_description`
