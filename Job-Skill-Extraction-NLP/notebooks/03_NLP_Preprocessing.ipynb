# Day 6 — NLP Text Preprocessing

Tokenization, lowercasing, stopword removal, lemmatization, stemming comparison, and sentence segmentation, applied to a sample job description.

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
import pandas as pd

df = pd.read_csv('clean_jobs.csv')
raw_sample = df['job_description'].iloc[0]
print(raw_sample)
```

## Sentence segmentation (on the original, punctuated text)

```python
sentences = sent_tokenize(raw_sample)
for s in sentences:
    print('-', s)
```

## Tokenization + lowercasing

```python
tokens = word_tokenize(raw_sample.lower())
print(tokens)
```

## Stopword removal

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [t for t in tokens if t.isalpha() and t not in stop_words]
print(filtered_tokens)
```

## Lemmatization vs Stemming comparison

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

comparison = pd.DataFrame({
    'token': filtered_tokens,
    'stemmed': [stemmer.stem(t) for t in filtered_tokens],
    'lemmatized': [lemmatizer.lemmatize(t) for t in filtered_tokens],
})
comparison.head(15)
```

## Apply the full pipeline to the whole dataset (tokens + stopword removal)

```python
def preprocess(text):
    tokens = word_tokenize(str(text).lower())
    tokens = [t for t in tokens if t.isalpha() and t not in stop_words]
    return tokens

df['tokens'] = df['clean_description'].apply(preprocess)
df['token_count'] = df['tokens'].apply(len)
df[['clean_description','tokens','token_count']].head(3)
```

## Summary
- Sentence segmentation, tokenization, lowercasing, stopword removal all applied
- Stemming (Porter) is faster but produces truncated/non-word forms (e.g. 'analyz')
- Lemmatization (WordNet) produces valid dictionary words, generally preferred for skill text
- Full dataset tokenized for the next steps (rule-based extraction, TF-IDF, n-grams)
