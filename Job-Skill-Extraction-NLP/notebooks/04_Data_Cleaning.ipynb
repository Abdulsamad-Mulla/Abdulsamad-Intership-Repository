# Day 4 — Data Cleaning

Cleans `job_description` text: removes HTML, URLs, emails, phone numbers, special characters, extra spaces; lowercases the text.

```python
import pandas as pd
import re

df = pd.read_csv('../data/jobs_skills_Extraction_10000.csv')

def clean_text(text):
    text = str(text)
    text = re.sub(r'http\S+', '', text)          # URLs
    text = re.sub(r'<.*?>', '', text)             # HTML tags
    text = re.sub(r'\S+@\S+', '', text)           # emails
    text = re.sub(r'\+?\d[\d\-\s]{7,}\d', '', text)  # phone numbers
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)   # special chars
    text = re.sub(r'\s+', ' ', text)              # extra spaces
    return text.strip().lower()

df['clean_description'] = df['job_description'].apply(clean_text)
df[['job_description','clean_description']].head(3)
```

## Before / after example

```python
print('BEFORE:\n', df['job_description'].iloc[0])
print('\nAFTER:\n', df['clean_description'].iloc[0])
```

## checks

```python
print('Any nulls in clean_description:', df['clean_description'].isnull().sum())
print('Avg length before:', df['job_description'].str.len().mean().round(1))
print('Avg length after :', df['clean_description'].str.len().mean().round(1))
```

## Save cleaned dataset

```python
df.to_csv('../data/clean_jobs.csv', index=False)
print('Saved clean_jobs.csv with', len(df), 'rows')
```

## Summary
- Applied regex-based cleaning: URLs, HTML, emails, phone numbers, special characters, extra whitespace removed
- Text lowercased for downstream NLP consistency
- New column `clean_description` added, saved alongside original columns
