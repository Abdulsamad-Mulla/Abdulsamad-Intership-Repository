```python
# Run a pretrained transformer NER model (first cell)

from transformers import pipeline

#  dslim/bert-base-NER is a well-known pretrained NER model
ner_pipeline = pipeline("ner", model = "dslim/bert-base-NER", aggregation_strategy="simple")

sample_text = "Experience with Python, SQL and AWS."
results = ner_pipeline(sample_text)

for r in results:
    print(f"{r["word"]} -> {r["entity_group"]} (score: {r["score"]:.2f})")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
if "entity_ruler" not in nlp.pipe_names:
    ruler = nlp.add_pipe("entity_ruler", before="ner")
    ruler.add_patterns([
        {"label": "SKILL", "pattern": "Python"},
        {"label": "DATABASE", "pattern": "SQL"},
        {"label": "CLOUD_PLATFORM", "pattern": "AWS"},
    ])

doc = nlp(sample_text)

print("Pretrained BERT NER:")
for r in results:
    print(f"  {r['word']} -> {r['entity_group']}")

print("\nCustom EntityRuler (Day 12 approach):")
for ent in doc.ents:
    print(f"  {ent.text} -> {ent.label_}")
```
