# Day 2 — Dataset Collection & Initial Analysis

**Dataset:** `raw_jobs.csv` — 10,000 job postings with fields: `job_id, job_title, company, location, job_description, experience, education, salary, job_type`.

This notebook inspects the dataset shape, columns, missing values, duplicates and text length, as required for Day 2.

```python
import pandas as pd

df = pd.read_csv('../data/jobs_skills_Extraction_10000.csv')
print(df.shape)
print(df.columns.tolist())
df.head()
```

## Missing values per column

```python
df.isnull().sum()
```

## Duplicate records

```python
print('Duplicate job_id:', df['job_id'].duplicated().sum())
print('Fully duplicate rows:', df.duplicated().sum())
```

## Text length of job_description

```python
df['description_length'] = df['job_description'].str.len()
df['description_length'].describe()
```

## Relevant columns for skill extraction
The most important column for skill extraction is `job_description`, since it contains free-text mentions of required skills. `job_title`, `experience`, and `education` provide useful context/metadata.

```python
df[['job_title','job_description']].sample(3, random_state=1)
```

## Summary
- 10,000 records, 9 columns
- No duplicate `job_id`s or fully duplicate rows
- Missing values only in `company`, `education`, `salary` (non-critical for skill extraction)
- `job_description` is present for all rows and is the key text field going forward
