````python
# Ner (named Entity Recognition)
import spacy
nlp = spacy.load("en_core_web_sm")

sample_text = "We need a candidate skilled in Python, Aws, PostgreSQL, and Power BI, based in New York."
doc = nlp(sample_text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
````

````python
if "entity_ruler" not in nlp.pipe_names:
    ruler = nlp.add_pipe("entity_ruler", before="ner")
    
patterns = [
    {"label": "SKILL", "pattern": "Python"},
    {"label": "SKILL", "pattern": "SQL"},

    {"label": "DATABASE", "pattern": "PostgreSQL"},
    {"label": "DATABASE", "pattern": "MySQL"},

    {"label": "TOOL", "pattern": "Power BI"},
    {"label": "TOOL", "pattern": "Tableau"},

    {"label": "CLOUD_PLATFORM", "pattern": "AWS"},
    {"label": "CLOUD_PLATFORM", "pattern": "Amazon Web Services"},
]
    
ruler.add_patterns(patterns)

doc2 = nlp(sample_text)

for ent in doc2.ents:
    print(ent.text, "->", ent.label_)
````
